```python
import pandas as pd
import numpy as np

# ============================================
# STEP 1: Load Cleaned Dataset
# ============================================

df = pd.read_csv("Cleaned_Sample_Dataset.csv")

print("Dataset loaded successfully!")
print("Shape:", df.shape)


# ============================================
# STEP 2: Display First 5 Rows
# ============================================

print("\nFirst 5 Rows:")
print(df.head())


# ============================================
# STEP 3: Check Dataset Information
# ============================================

print("\nDataset Information:")
print(df.info())


# ============================================
# STEP 4: Check Column Names
# ============================================

print("\nColumn Names:")
print(df.columns.tolist())


# ============================================
# STEP 5: Check Missing Values
# ============================================

print("\nMissing Values:")
print(df.isnull().sum())


# ============================================
# STEP 6: Check Duplicate Values
# ============================================

print("\nDuplicate Rows:", df.duplicated().sum())


# ============================================
# STEP 7: Statistical Summary
# ============================================

print("\nStatistical Summary:")
print(df.describe())


# ============================================
# STEP 8: Select Numeric Columns
# ============================================

numeric_df = df.select_dtypes(include=np.number)

print("\nNumeric Columns:")
print(numeric_df.columns.tolist())


# ============================================
# STEP 9: Calculate Correlation Matrix
# ============================================

correlation_matrix = numeric_df.corr()

print("\nCorrelation Matrix:")
print(correlation_matrix.round(2))


# ============================================
# STEP 10: Find Strong Correlations
# ============================================

corr_pairs = correlation_matrix.unstack()

corr_pairs = corr_pairs[
    (corr_pairs != 1)
]

corr_pairs = corr_pairs.sort_values(
    key=lambda x: abs(x),
    ascending=False
)

print("\nStrongest Correlations:")
print(corr_pairs.head(10).round(2))


# ============================================
# STEP 11: Find Positive Correlations
# ============================================

positive_corr = corr_pairs[corr_pairs > 0]

print("\nTop Positive Correlations:")
print(positive_corr.head(10).round(2))


# ============================================
# STEP 12: Find Negative Correlations
# ============================================

negative_corr = corr_pairs[corr_pairs < 0]

print("\nTop Negative Correlations:")
print(negative_corr.head(10).round(2))


# ============================================
# STEP 13: Check Correlation with Revenue
# ============================================

if "Revenue" in correlation_matrix.columns:

    revenue_corr = correlation_matrix["Revenue"].sort_values(
        ascending=False
    )

    print("\nCorrelation with Revenue:")
    print(revenue_corr.round(2))


# ============================================
# STEP 14: Check Correlation with Profit
# ============================================

if "Profit" in correlation_matrix.columns:

    profit_corr = correlation_matrix["Profit"].sort_values(
        ascending=False
    )

    print("\nCorrelation with Profit:")
    print(profit_corr.round(2))


# ============================================
# STEP 15: Identify Strong Relationships
# ============================================

strong_relationships = correlation_matrix[
    (correlation_matrix.abs() >= 0.70) &
    (correlation_matrix.abs() < 1.00)
]

print("\nStrong Relationships (|Correlation| >= 0.70):")
print(strong_relationships.round(2))


# ============================================
# STEP 16: Final Correlation Matrix
# ============================================

final_output = correlation_matrix.round(2)


# ============================================
# STEP 17: Save Final Output
# ============================================

output_file = "Correlation_Analysis.csv"

final_output.to_csv(output_file)

print("\nAnalysis completed successfully!")
print("Output file:", output_file)
```
